```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor
)
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage
import statsmodels.api as sm

# ================================
# EDA - GDP Data Analysis
# ================================

### Kaggle paths ###

DATA_PATH = "/kaggle/input/globalmigrationpattern/API_NY.GDP.PCAP.CD_DS2_en_csv_v2_24794.csv"
PLOTS_DIR = "/kaggle/working/plots"

# Create plots directory
os.makedirs(PLOTS_DIR, exist_ok=True)


### Load Data ###

try:
    df = pd.read_csv(DATA_PATH, skiprows=4)
except Exception as e:
    raise RuntimeError(f"Error reading GDP data: {e}")

### Data Cleaning ###

df = df.drop(columns=["Indicator Name", "Indicator Code"], errors="ignore")

# Convert from wide to long format
df = df.melt(
    id_vars=["Country Name", "Country Code"],
    var_name="Year",
    value_name="GDP_per_capita"
)

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df = df.dropna(subset=["GDP_per_capita"])
df["GDP_per_capita"] = df["GDP_per_capita"].astype(float)

print(f"Cleaned GDP data shape: {df.shape}")

### Descriptive Statistics ###

print("\n=== GDP per Capita Summary (first 5 years) ===")
display(df.groupby("Year")["GDP_per_capita"].describe().head())

### Visualization Settings ###

sns.set(style="whitegrid")

### Global GDP Trend Over Time ###

plt.figure(figsize=(10, 5))
global_mean = df.groupby("Year")["GDP_per_capita"].mean()
sns.lineplot(x=global_mean.index, y=global_mean.values)

plt.title("Average Global GDP per Capita Over Time")
plt.xlabel("Year")
plt.ylabel("GDP per Capita (USD)")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "global_gdp_trend.png"))
plt.show()

### Top 10 Richest Countries (Latest Year) ###

latest_year = int(df["Year"].max())

top10 = (
    df[df["Year"] == latest_year]
    .nlargest(10, "GDP_per_capita")
    .set_index("Country Name")["GDP_per_capita"]
)

plt.figure(figsize=(8, 5))
sns.barplot(x=top10.values, y=top10.index, palette="crest")

plt.title(f"Top 10 Countries by GDP per Capita ({latest_year})")
plt.xlabel("GDP per Capita (USD)")
plt.ylabel("")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "top10_gdp_countries.png"))
plt.show()

### GDP Distribution (Latest Year) ###

plt.figure(figsize=(8, 5))
sns.histplot(
    df[df["Year"] == latest_year]["GDP_per_capita"],
    bins=40,
    kde=True
)

plt.title(f"GDP per Capita Distribution ({latest_year})")
plt.xlabel("GDP per Capita (USD)")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "gdp_distribution.png"))
plt.show()

print(f"Plots saved in: {PLOTS_DIR}")


# ======================================
# EDA - HDI Trends Analysis
# ======================================

### Paths (Kaggle) ###

DATA_PATH = "/kaggle/input/globalmigrationpattern/HDR25_Statistical_Annex_HDI_Trends_Table.xlsx"
PLOTS_DIR = "/kaggle/working/plots"

os.makedirs(PLOTS_DIR, exist_ok=True)

### Automatically Detect Correct Sheet ###

xls = pd.ExcelFile(DATA_PATH)
sheet_name = [s for s in xls.sheet_names if "HDI" in s or "Table 2" in s][0]
print(f"Using sheet: {sheet_name}")

### Read Excel & Auto-Detect Header ###

df_raw = pd.read_excel(DATA_PATH, sheet_name=sheet_name, header=None)

header_row = None
for i in range(10):
    row_values = df_raw.iloc[i].astype(str).values
    if "Country" in row_values or "HDI rank" in row_values:
        header_row = i
        break

if header_row is None:
    raise ValueError("Could not detect header row automatically.")

df = pd.read_excel(DATA_PATH, sheet_name=sheet_name, header=header_row)
df = df.rename(columns=lambda x: str(x).strip())

### Identify Country Column ###

country_col = [c for c in df.columns if "Country" in c or "country" in c.lower()]
if not country_col:
    raise ValueError("Could not find Country column.")
country_col = country_col[0]

### Keep Relevant Year Columns ###

years = [str(y) for y in range(1990, 2024) if str(y) in df.columns]
df = df[[country_col] + years]

### Data Cleaning ###

df[years] = df[years].apply(pd.to_numeric, errors="coerce")
df = df.dropna(subset=[country_col])
df = df[df[years].notna().any(axis=1)]

# Remove regional/group aggregates
df = df[~df[country_col].str.contains("development", case=False, na=False)]

df = df.set_index(country_col)

print(f"Cleaned data shape: {df.shape}")


### Summary Preview ###

print("\n=== HDI Summary (first 5 rows) ===")
display(df.head())

### Visualization Settings ###

sns.set(style="whitegrid")

### Plot 1: Global Average HDI Trend ###

plt.figure(figsize=(10, 5))
df[years].mean().plot(marker="o")

plt.title("Average Global HDI Trend (1990–2023)")
plt.xlabel("Year")
plt.ylabel("Average HDI")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "hdi_global_trend.png"))
plt.show()

### Plot 2: Top 10 Countries (2023) ###

top10 = df["2023"].sort_values(ascending=False).head(10)

plt.figure(figsize=(8, 5))
sns.barplot(x=top10.values, y=top10.index, palette="crest")

plt.title("Top 10 Countries by HDI (2023)")
plt.xlabel("HDI Value")
plt.ylabel("")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "hdi_top10_2023.png"))
plt.show()

### Plot 3: Bottom 10 Countries (2023) ###

bottom10 = df["2023"].sort_values().head(10)

plt.figure(figsize=(8, 5))
sns.barplot(x=bottom10.values, y=bottom10.index, palette="rocket")

plt.title("Bottom 10 Countries by HDI (2023)")
plt.xlabel("HDI Value")
plt.ylabel("")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "hdi_bottom10_2023.png"))
plt.show()

### Plot 4: Correlation Heatmap ###

plt.figure(figsize=(10, 7))
sns.heatmap(
    df[years].corr(),
    cmap="coolwarm",
    fmt=".2f",
    cbar_kws={"label": "Correlation"}
)

plt.title("Correlation Between HDI Across Years")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "hdi_correlation_heatmap.png"))
plt.show()

print(f"All plots saved to: {PLOTS_DIR}")


# ==================================================
# EDA - UN DESA International Migration
# ==================================================

### Paths (Kaggle) ###

DATA_PATH = "/kaggle/input/globalmigrationpattern/undesa_pd_2024_ims_stock_by_sex_destination_and_origin.xlsx"
PLOTS_DIR = "/kaggle/working/plots"

os.makedirs(PLOTS_DIR, exist_ok=True)

### Load Data ###

xls = pd.ExcelFile(DATA_PATH)
df = pd.read_excel(xls, sheet_name="Table 1", header=10)

### Keep Relevant Columns ###

df = df[
    [
        'Region, development group, country or area of destination',
        'Region, development group, country or area of origin',
        1990, 1995, 2000, 2005, 2010, 2015, 2020, 2024
    ]
]

df.columns = [
    'Destination', 'Origin',
    '1990', '1995', '2000', '2005',
    '2010', '2015', '2020', '2024'
]

print(f"Data Loaded: {df.shape[0]} rows, {df.shape[1]} columns")

### Visualization Settings ###

sns.set(style="whitegrid")

### Global Migration Trend ###

df_years = df.loc[:, '1990':'2024'].sum()

plt.figure(figsize=(8, 5))
sns.lineplot(x=df_years.index, y=df_years.values, marker="o")

plt.title("Global International Migrant Stock (1990–2024)")
plt.xlabel("Year")
plt.ylabel("Total Migrants")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "migration_global_trend.png"))
plt.show()

### Top 10 Destination Countries (2024) ###

top_dest = (
    df.groupby("Destination")["2024"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8, 5))
sns.barplot(x=top_dest.values, y=top_dest.index, palette="crest")

plt.title("Top 10 Destination Countries (2024)")
plt.xlabel("Migrant Stock")
plt.ylabel("")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "migration_top_destinations_2024.png"))
plt.show()

### Top 10 Origin Countries (2024) ###

top_orig = (
    df.groupby("Origin")["2024"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8, 5))
sns.barplot(x=top_orig.values, y=top_orig.index, palette="mako")

plt.title("Top 10 Origin Countries (2024)")
plt.xlabel("Migrant Stock")
plt.ylabel("")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "migration_top_origins_2024.png"))
plt.show()

print(f"All plots saved to: {PLOTS_DIR}")


# ==================================================
# EDA - Urban Population (% of Total)
# ==================================================

### Paths (Kaggle) ###

DATA_PATH = "/kaggle/input/globalmigrationpattern/API_SP.URB.TOTL.IN.ZS_DS2_en_csv_v2_129596.csv"
PLOTS_DIR = "/kaggle/working/plots"

### Load Dataset ###

df = pd.read_csv(DATA_PATH, skiprows=4)

# Drop unnamed columns (metadata leftovers)
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

print(f"Dataset shape after dropping extra columns: {df.shape}")
display(df.head())

### Data Cleaning ###

df_clean = df.drop(columns=["Indicator Name", "Indicator Code"], errors="ignore")
df_clean = df_clean.rename(
    columns={"Country Name": "Country", "Country Code": "Code"}
)

# Melt to long format
df_long = df_clean.melt(
    id_vars=["Country", "Code"],
    var_name="Year",
    value_name="Urban_Pop_Percent"
)

# Convert Year safely to integer
df_long = df_long[pd.to_numeric(df_long["Year"], errors="coerce").notnull()]
df_long["Year"] = df_long["Year"].astype(int)

# Convert values to numeric
df_long["Urban_Pop_Percent"] = pd.to_numeric(
    df_long["Urban_Pop_Percent"], errors="coerce"
)

print("Long-format data preview:")
display(df_long.head())

os.makedirs(PLOTS_DIR, exist_ok=True)

### Top & Bottom Urbanized Countries (2024) ###

latest_year = df_long[df_long["Year"] == 2024].sort_values(
    by="Urban_Pop_Percent", ascending=False
)

print("Top 10 urbanized countries in 2024:")
display(latest_year[["Country", "Urban_Pop_Percent"]].head(10))

print("Bottom 10 urbanized countries in 2024:")
display(latest_year[["Country", "Urban_Pop_Percent"]].tail(10))

### Global Urbanization Trend ###

plt.figure(figsize=(14, 6))
sns.lineplot(
    data=df_long.groupby("Year")["Urban_Pop_Percent"].mean().reset_index(),
    x="Year",
    y="Urban_Pop_Percent"
)

plt.title("Global Average Urbanization Over Time")
plt.ylabel("Urban population (% of total)")
plt.xlabel("Year")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "global_avg_urban.png"))
plt.show()

### Urbanization Trends (Top Countries) ###

top_countries = latest_year["Country"].head(10)

plt.figure(figsize=(16, 8))
sns.lineplot(
    data=df_long[df_long["Country"].isin(top_countries)],
    x="Year",
    y="Urban_Pop_Percent",
    hue="Country",
    marker="o"
)

plt.title("Urbanization Trends for Top Countries (2024)")
plt.ylabel("Urban population (% of total)")
plt.xlabel("Year")
plt.legend(loc="lower right")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "urban_trend_top.png"))
plt.show()

### Distribution of Urbanization (2024) ###

plt.figure(figsize=(12, 6))
sns.histplot(latest_year["Urban_Pop_Percent"], bins=30, kde=True)

plt.title("Distribution of Urbanization in 2024")
plt.xlabel("Urban population (% of total)")
plt.ylabel("Number of countries")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "urbanization_in_2024.png"))
plt.show()

### Heatmap: Selected Countries Over Time ###

selected_countries = df_long["Country"].unique()[:20]

pivot_df = (
    df_long[df_long["Country"].isin(selected_countries)]
    .pivot(index="Country", columns="Year", values="Urban_Pop_Percent")
)

plt.figure(figsize=(18, 10))
sns.heatmap(pivot_df, cmap="YlGnBu")

plt.title("Urbanization % Over Time for Selected Countries")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "urban_percent_selected_countries.png"))
plt.show()

print(f"All plots saved to: {PLOTS_DIR}")


# ==================================================
# Merging Datasets
# (Migration + GDP + HDI + Urbanization)
# ==================================================

### Paths (Kaggle) ###

DATA_DIR = "/kaggle/input/globalmigrationpattern"
OUTPUT_DIR = "/kaggle/working/processed"
os.makedirs(OUTPUT_DIR, exist_ok=True)

### Load Migration Data ###

print("Loading Migration Data...")
migration_path = os.path.join(
    DATA_DIR,
    "undesa_pd_2024_ims_stock_by_sex_destination_and_origin.xlsx"
)

xls = pd.ExcelFile(migration_path)
migration_df = pd.read_excel(xls, sheet_name="Table 1", header=10)

migration_df = migration_df[
    [
        'Region, development group, country or area of destination',
        'Region, development group, country or area of origin',
        1990, 1995, 2000, 2005, 2010, 2015, 2020, 2024
    ]
]

migration_df.columns = [
    'Destination', 'Origin',
    '1990', '1995', '2000', '2005',
    '2010', '2015', '2020', '2024'
]

migration_df = migration_df.rename(columns={'Destination': 'Country'})
migration_df = migration_df.drop(columns=['Origin'])

# Convert to long format
migration_long = migration_df.melt(
    id_vars=['Country'],
    var_name='Year',
    value_name='Migration'
)
migration_long['Year'] = pd.to_numeric(migration_long['Year'])

print(f"Migration data shape: {migration_long.shape}")

### Load GDP Data ###

print("Loading GDP Data...")
gdp_path = os.path.join(
    DATA_DIR,
    "API_NY.GDP.PCAP.CD_DS2_en_csv_v2_24794.csv"
)

gdp_df = pd.read_csv(gdp_path, skiprows=4)
gdp_df = gdp_df.drop(columns=["Indicator Name", "Indicator Code"], errors="ignore")

gdp_df = gdp_df.melt(
    id_vars=["Country Name", "Country Code"],
    var_name="Year",
    value_name="GDP_per_capita"
)

gdp_df["Year"] = pd.to_numeric(gdp_df["Year"], errors="coerce")
gdp_df = gdp_df.dropna(subset=["GDP_per_capita", "Year"])
gdp_df = gdp_df.rename(columns={"Country Name": "Country"})

print(f"GDP data shape: {gdp_df.shape}")

### Load Urbanization Data ###

print("Loading Urbanization Data...")
urb_path = os.path.join(
    DATA_DIR,
    "API_SP.URB.TOTL.IN.ZS_DS2_en_csv_v2_129596.csv"
)

urb_df = pd.read_csv(urb_path, skiprows=4)
urb_df = urb_df.drop(columns=["Indicator Name", "Indicator Code"], errors="ignore")

urb_df = urb_df.melt(
    id_vars=["Country Name", "Country Code"],
    var_name="Year",
    value_name="Urbanization"
)

urb_df["Year"] = pd.to_numeric(urb_df["Year"], errors="coerce")
urb_df = urb_df.dropna(subset=["Urbanization", "Year"])
urb_df = urb_df.rename(columns={"Country Name": "Country"})

print(f"Urbanization data shape: {urb_df.shape}")

### Load HDI Data ###

print("Loading HDI Data...")
hdi_path = os.path.join(
    DATA_DIR,
    "HDR25_Statistical_Annex_HDI_Trends_Table.xlsx"
)

xls = pd.ExcelFile(hdi_path)
sheet_name = [s for s in xls.sheet_names if "HDI" in s or "Table 2" in s][0]

hdi_raw = pd.read_excel(hdi_path, sheet_name=sheet_name, header=None)

header_row = None
for i in range(10):
    if "Country" in hdi_raw.iloc[i].astype(str).values:
        header_row = i
        break

hdi_df = pd.read_excel(hdi_path, sheet_name=sheet_name, header=header_row)
hdi_df.columns = hdi_df.columns.map(str)

country_col = [c for c in hdi_df.columns if "Country" in c][0]
years = [str(y) for y in range(1990, 2024) if str(y) in hdi_df.columns]

hdi_df = hdi_df[[country_col] + years]
hdi_df = hdi_df.rename(columns={country_col: "Country"})

hdi_df = hdi_df.melt(
    id_vars=["Country"],
    var_name="Year",
    value_name="HDI"
)

hdi_df["Year"] = pd.to_numeric(hdi_df["Year"], errors="coerce")
hdi_df = hdi_df.dropna(subset=["HDI"])

print(f"HDI data shape: {hdi_df.shape}")

### Clean Country Names ###

def clean_country_name(name):
    if isinstance(name, str):
        return name.strip().replace("*", "").replace("...", "")
    return name

for df in [migration_long, gdp_df, hdi_df, urb_df]:
    df["Country"] = df["Country"].apply(clean_country_name)

### Merge Datasets ###

print("Merging datasets...")

merged_df = migration_long.merge(
    gdp_df, on=["Country", "Year"], how="inner"
)

merged_df = merged_df.merge(
    hdi_df, on=["Country", "Year"], how="inner"
)

merged_df = merged_df.merge(
    urb_df, on=["Country", "Year"], how="inner"
)

print(f"Merged data shape: {merged_df.shape}")

### Remove Missing Values ###

missing_before = merged_df.isna().sum().sum()

merged_df = merged_df.dropna(
    subset=["Migration", "GDP_per_capita", "HDI", "Urbanization"]
)

missing_after = merged_df.isna().sum().sum()
print(f"Missing values removed: {missing_before - missing_after}")

### Save Final Dataset ###

output_file = os.path.join(
    OUTPUT_DIR,
    "merged_global_migration_data.csv"
)

merged_df.to_csv(output_file, index=False)

print(f"Merged dataset saved at: {output_file}")


# ==================================================
# Combined EDA
# (Migration, GDP, HDI, Urbanization)
# ==================================================

### Paths (Kaggle) ###

DATA_PATH = "/kaggle/working/processed/merged_global_migration_data.csv"
PLOTS_DIR = "/kaggle/working/plots"

os.makedirs(PLOTS_DIR, exist_ok=True)

### Load Data ###

df = pd.read_csv(DATA_PATH)
print(f"Loaded merged dataset: {df.shape[0]} rows")

### Force Numeric Conversion ###

for col in ["Migration", "GDP_per_capita", "HDI", "Urbanization"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Drop rows with missing values
df = df.dropna(subset=["Migration", "GDP_per_capita", "HDI", "Urbanization"])
print(f"After cleaning: {df.shape[0]} rows remain")

### Aggregate Migration per Country-Year ###

df_agg = (
    df.groupby(["Country", "Year"], as_index=False)
    .agg(
        Migration=("Migration", "sum"),
        GDP_per_capita=("GDP_per_capita", "mean"),
        HDI=("HDI", "mean"),
        Urbanization=("Urbanization", "mean"),
    )
)

print(f"After aggregation: {df_agg.shape[0]} rows")

### Correlation Matrix ###

corr = df_agg[
    ["Migration", "GDP_per_capita", "HDI", "Urbanization"]
].corr()

print("\n=== Correlation Matrix ===")
display(corr)

plt.figure(figsize=(7, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Between Migration, GDP, HDI, and Urbanization")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "correlation_heatmap.png"))
plt.show()

### Scatter Plots ###

for col in ["GDP_per_capita", "HDI", "Urbanization"]:
    plt.figure(figsize=(7, 5))
    sns.scatterplot(
        x=col,
        y="Migration",
        data=df_agg,
        alpha=0.6
    )

    plt.title(f"Migration vs {col}")
    plt.xlabel(col)
    plt.ylabel("Migration Stock")
    plt.tight_layout()

    plt.savefig(
        os.path.join(PLOTS_DIR, f"migration_vs_{col.lower()}.png")
    )
    plt.show()

### Yearly Migration Trend ###

plt.figure(figsize=(8, 5))
sns.lineplot(
    data=df_agg.groupby("Year")["Migration"].mean().reset_index(),
    x="Year",
    y="Migration"
)

plt.title("Average Global Migration Trend (1990–2024)")
plt.ylabel("Mean Migration Stock")
plt.xlabel("Year")
plt.tight_layout()

plt.savefig(os.path.join(PLOTS_DIR, "global_trends.png"))
plt.show()

### Output Confirmation ###

print(f"All plots saved to: {PLOTS_DIR}")


# ==================================================
# Statistical & ML Modeling
# Migration Drivers Analysis
# ==================================================

### Paths (Kaggle) ###

DATA_PATH = "/kaggle/working/processed/merged_global_migration_data.csv"
PLOTS_DIR = "/kaggle/working/plots"
OUTPUT_DIR = "/kaggle/working/processed"

os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)

### Load Cleaned Data ###

df = pd.read_csv(DATA_PATH)
print(f"Loaded merged dataset: {df.shape[0]} rows")

# Convert to numeric & drop missing
for col in ["Migration", "GDP_per_capita", "HDI", "Urbanization"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["Migration", "GDP_per_capita", "HDI", "Urbanization"])

### Aggregate Data (Country-Level Means) ###

df_country = (
    df.groupby("Country", as_index=False)
    .agg(
        Migration=("Migration", "mean"),
        GDP_per_capita=("GDP_per_capita", "mean"),
        HDI=("HDI", "mean"),
        Urbanization=("Urbanization", "mean"),
    )
)

print(f"Aggregated dataset: {df_country.shape[0]} countries")

### 1. OLS REGRESSION ###

print("\nRunning OLS Regression...")

X = df_country[["GDP_per_capita", "HDI", "Urbanization"]]
y = df_country["Migration"]

X = sm.add_constant(X)
ols_model = sm.OLS(y, X).fit()
print(ols_model.summary())

# Regression plots
for col in ["GDP_per_capita", "HDI", "Urbanization"]:
    plt.figure(figsize=(6, 4))
    sns.regplot(
        x=col,
        y="Migration",
        data=df_country,
        scatter_kws={"alpha": 0.6}
    )
    plt.title(f"Regression: Migration vs {col}")
    plt.tight_layout()
    plt.savefig(os.path.join(PLOTS_DIR, f"regression_migration_{col.lower()}.png"))
    plt.show()


### 2. RANDOM FOREST FEATURE IMPORTANCE ###

print("\nRunning Random Forest...")

rf_model = RandomForestRegressor(n_estimators=300, random_state=42)
rf_model.fit(X.drop(columns="const"), y)

rf_imp = pd.DataFrame({
    "Feature": X.drop(columns="const").columns,
    "Importance": rf_model.feature_importances_
})

print("\nRandom Forest – Feature Importance:")
display(rf_imp.sort_values(by="Importance", ascending=False))

plt.figure(figsize=(6, 4))
sns.barplot(data=rf_imp, x="Feature", y="Importance")
plt.title("Random Forest Feature Importance")
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "rf_feature_importance.png"))
plt.show()

### 2B. EXTRA TREES

print("\nRunning Extra Trees...")

et_model = ExtraTreesRegressor(n_estimators=300, random_state=42)
et_model.fit(X.drop(columns="const"), y)

et_imp = pd.DataFrame({
    "Feature": X.drop(columns="const").columns,
    "Importance": et_model.feature_importances_
})

print("\nExtra Trees – Feature Importance:")
display(et_imp.sort_values(by="Importance", ascending=False))

plt.figure(figsize=(6, 4))
sns.barplot(data=et_imp, x="Feature", y="Importance")
plt.title("Extra Trees Feature Importance")
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "extratrees_feature_importance.png"))
plt.show()

### 2C. GRADIENT BOOSTING ###

print("\nRunning Gradient Boosting...")

gb_model = GradientBoostingRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)
gb_model.fit(X.drop(columns="const"), y)

gb_imp = pd.DataFrame({
    "Feature": X.drop(columns="const").columns,
    "Importance": gb_model.feature_importances_
})

print("\nGradient Boosting – Feature Importance:")
display(gb_imp.sort_values(by="Importance", ascending=False))

plt.figure(figsize=(6, 4))
sns.barplot(data=gb_imp, x="Feature", y="Importance")
plt.title("Gradient Boosting Feature Importance")
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "gradientboosting_feature_importance.png"))
plt.show()

### 3. K-MEANS CLUSTERING ###

print("\nRunning K-Means Clustering...")

features = df_country[["Migration", "GDP_per_capita", "HDI", "Urbanization"]]
scaler = StandardScaler()
scaled = scaler.fit_transform(features)

kmeans = KMeans(n_clusters=3, random_state=42)
df_country["KMeans_Cluster"] = kmeans.fit_predict(scaled)

silhouette = silhouette_score(scaled, df_country["KMeans_Cluster"])
print("K-Means Silhouette Score:", silhouette)

print("\nCluster Means:")
display(
    df_country.groupby("KMeans_Cluster")[features.columns].mean()
)

# Visualizations
plt.figure(figsize=(7, 5))
sns.scatterplot(
    x="GDP_per_capita",
    y="Migration",
    hue="KMeans_Cluster",
    data=df_country,
    palette="viridis",
    s=60
)
plt.title("K-Means Clusters: GDP vs Migration")
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "kmeans_clusters_gdp_migration.png"))
plt.show()

plt.figure(figsize=(7, 5))
sns.scatterplot(
    x="Urbanization",
    y="Migration",
    hue="KMeans_Cluster",
    data=df_country,
    palette="viridis",
    s=60
)
plt.title("K-Means Clusters: Urbanization vs Migration")
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "kmeans_clusters_urbanization_migration.png"))
plt.show()

### 4. HIERARCHICAL CLUSTERING ###

print("\nRunning Hierarchical Clustering...")

linked = linkage(scaled, method="ward")

plt.figure(figsize=(20, 8))
dendrogram(
    linked,
    labels=df_country["Country"].values,
    leaf_rotation=90,
    leaf_font_size=10,
    color_threshold=0.7 * linked[:, 2].max()
)
plt.title("Hierarchical Clustering (Ward)")
plt.xlabel("Countries")
plt.ylabel("Distance")
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, "hierarchical_dendrogram.png"), dpi=300)
plt.show()

hc = AgglomerativeClustering(n_clusters=3, linkage="ward")
df_country["HierCluster"] = hc.fit_predict(scaled)

print("\nHierarchical Cluster Means:")
display(
    df_country.groupby("HierCluster")[features.columns].mean()
)

### Save Results ###

output_file = os.path.join(
    OUTPUT_DIR,
    "model_results_country_level.csv"
)

df_country.to_csv(output_file, index=False)

print("\nAll modeling complete!")
print(f"Plots saved in: {PLOTS_DIR}")
print(f"Model results saved at: {output_file}")


# ======================================================
# Feature Importance & Cluster Comparison
# ======================================================

### Paths ###

PLOTS_DIR = "/kaggle/working/plots"
os.makedirs(PLOTS_DIR, exist_ok=True)

### Feature Importance Data ###

feature_importance = pd.DataFrame({
    "Feature": ["GDP_per_capita", "HDI", "Urbanization"],
    "Random Forest": [0.341902, 0.421346, 0.236752],
    "Extra Trees": [0.469544, 0.268559, 0.261897],
    "Gradient Boosting": [0.275687, 0.394275, 0.330038]
})

print("Feature Importance Data:")
display(feature_importance)

### Cluster Means Data ###

cluster_means = pd.DataFrame({
    "Cluster": ["KMeans 0", "KMeans 1", "KMeans 2",
                "Hier 0", "Hier 1", "Hier 2"],
    "Migration": [129920, 582617, 102628, 341686, 259538, 120525],
    "GDP_per_capita": [1353, 41823, 7642, 1025, 46111, 7416],
    "HDI": [0.511, 0.878, 0.740, 0.472, 0.894, 0.723],
    "Urbanization": [32.7, 80.3, 64.4, 31.0, 81.3, 60.6]
})

print("Cluster Means Data:")
display(cluster_means)

### Plotting ###

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

### Left Plot: Feature Importance Comparison ###

x = np.arange(len(feature_importance))
width = 0.25

axes[0].bar(x - width, feature_importance["Random Forest"], width, label="Random Forest")
axes[0].bar(x, feature_importance["Extra Trees"], width, label="Extra Trees")
axes[0].bar(x + width, feature_importance["Gradient Boosting"], width, label="Gradient Boosting")

axes[0].set_xticks(x)
axes[0].set_xticklabels(feature_importance["Feature"])
axes[0].set_ylabel("Feature Importance")
axes[0].set_title("Regression Feature Importance Comparison")
axes[0].legend()
axes[0].grid(axis="y", linestyle="--", alpha=0.7)

### Right Plot: Cluster Summary ###

cluster_means_plot = cluster_means.set_index("Cluster")

cluster_means_plot[
    ["Migration", "GDP_per_capita", "HDI", "Urbanization"]
].plot(
    kind="bar",
    ax=axes[1],
    width=0.7
)

axes[1].set_title("Cluster Summary (K-Means & Hierarchical)")
axes[1].set_ylabel("Value")
axes[1].grid(axis="y", linestyle="--", alpha=0.7)

### Save & Show ###

plt.tight_layout()
plt.savefig(f"{PLOTS_DIR}/comparison_feature_cluster.png", dpi=300)
plt.show()

print("Plot saved to:", f"{PLOTS_DIR}/comparison_feature_cluster.png")
```
